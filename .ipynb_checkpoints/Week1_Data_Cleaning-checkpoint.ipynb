#### Save Before Statistics

```python
rows_before = df.shape[0]
cols_before = df.shape[1]

null_before = df.isnull().sum()

duplicate_before = df.duplicated().sum()
```

#### Remove Duplicates

```python
# Duplicate rows provide no additional information,
# so removing them avoids double-counting.
df = df.drop_duplicates()
```

```python
# Fill numeric columns with median
df["Aging"] = df["Aging"].fillna(df["Aging"].median())
df["Sales"] = df["Sales"].fillna(df["Sales"].median())
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Discount"] = df["Discount"].fillna(df["Discount"].median())
df["Shipping_Cost"] = df["Shipping_Cost"].fillna(df["Shipping_Cost"].median())

# Fill categorical column with mode (most frequent value)
df["Order_Priority"] = df["Order_Priority"].fillna(df["Order_Priority"].mode()[0])
```

### Check again

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

### Save After Statistics

```python
rows_after = df.shape[0]
cols_after = df.shape[1]

null_after = df.isnull().sum()

duplicate_after = df.duplicated().sum()
```

#### Compare Before & After

```python
print("Rows Before:", rows_before)
print("Rows After :", rows_after)

print("Duplicates Removed:", duplicate_before)

print("\nNull Values Before")
print(null_before)

print("\nNull Values After")
print(null_after)
```

#### After cleaning the data
