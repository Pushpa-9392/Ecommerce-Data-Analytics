## Visulization

#### Sales by product category

```python
import matplotlib.pyplot as plt

# Calculate total sales for each category
category_sales = df.groupby("Product_Category")["Sales"].sum().sort_values(ascending=False)

# Create bar chart
plt.figure(figsize=(10,6))
category_sales.plot(kind="bar", color="purple", edgecolor="black")

# Add title and labels
plt.title("Total Sales by Product Category", fontsize=14)
plt.xlabel("Product Category", fontsize=12)
plt.ylabel("Total Sales", fontsize=12)

# Rotate x-axis labels
plt.xticks(rotation=360, ha="right")

plt.show()
```

#### Total Profit by product Category

```python
# calculate total proit by each category
total_profit_by_category=df.groupby('Product_Category')['Profit'].sum().sort_values(ascending=True)

# Create bar chart
plt.figure(figsize=(10,6))
total_profit_by_category.plot(kind="bar",color="Purple",edgecolor='Black')

# Add title and labels
plt.title("Total profit by product category",fontsize=14)
plt.xlabel("Total Product_category",fontsize=12)
plt.ylabel("Total Profit",fontsize=12)

# Rotate x_axis labels
plt.xticks(rotation=360,ha='right')
plt.show()
```

### Sales Vs Profit

```python
plt.figure(figsize=(10,4))
plt.scatter(df['Sales'],df['Profit'],color="Purple")
plt.title("Sales Vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

### Histogram-Distribution of sales

```python
plt.figure(figsize=(8,7))

# Create histogram
plt.hist(df["Sales"],color="Purple",edgecolor="black",bins=20)

# Create labels
plt.title("Distribution of sales",fontsize=14)
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

### Pie Chart - Payment method

```python
plt.figure(figsize=(5,5))
payment=df["Payment_method"].value_counts()
plt.pie(payment,labels=payment.index,autopct="%1.1f%%",
        startangle=90)
plt.show()
```

```python
import seaborn as sns
plt.figure(figsize=(6,4))

sns.heatmap(df.select_dtypes(include="number").corr(),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], dayfirst=True)

sales = df.groupby("Order_Date")["Sales"].sum()

plt.figure(figsize=(12,5))

sales.plot()

plt.title("Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.show()
```
